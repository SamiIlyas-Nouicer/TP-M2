```python
import os
from collections import Counter
import re
import math

def calculate_tf_idf(folder_path):
    token_counts_per_file = []
    document_frequencies = Counter()
    total_documents = 0

    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        if os.path.isfile(file_path):
            total_documents += 1
            with open(file_path, 'r', encoding='utf-8') as file:
                text = file.read()
                # Tokenization: Using regex to split on non-word characters
                tokens = re.findall(r'\w+', text.lower())
                token_count = Counter(tokens)
                max_frequency = max(token_count.values())

                # Save token count and document frequencies for TF and IDF
                token_counts_per_file.append({
                    'file': filename,
                    'token_count': token_count,
                    'max_frequency': max_frequency
                })

                # Update document frequency for each token (i.e., number of documents that contain the word)
                for token in set(tokens):  # Use `set` to avoid multiple counts in the same document
                    document_frequencies[token] += 1

    # Calculate TF-IDF for each word in each document
    tf_idf_scores = {}

    for file_data in token_counts_per_file:
        file_name = file_data['file']
        tf_idf_scores[file_name] = {}

        for token, count in file_data['token_count'].items():
            # Calculate TF (Term Frequency)
            tf = count / file_data['max_frequency']

            # Calculate IDF (Inverse Document Frequency)
            idf = math.log10(total_documents / document_frequencies[token])

            # Calculate TF-IDF
            tf_idf_scores[file_name][token] = tf * idf

    return tf_idf_scores

# Specify the folder path

# Calculate TF-IDF across all files in the folder
tf_idf_scores = calculate_tf_idf(folder_path)

# Display TF-IDF results
for file, tokens in tf_idf_scores.items():
    print(f"TF-IDF scores for file: {file}")
    for token, score in tokens.items():
        print(f"{token}: {score:.6f}")
    print()  # Blank line between files
```

```python
import pandas as pd
import numpy as np
from collections import defaultdict
import os
import math

# Define file paths for each preprocessing type CSV
file_paths = {
    "split": "path/to/split.csv",
    "splitporter": "path/to/splitporter.csv",
    "splitlancaster": "path/to/splitlancaster.csv",
    "tokenize": "path/to/tokenize.csv",
    "tokenizeporter": "path/to/tokenizeporter.csv",
    "tokenizelancaster": "path/to/tokenizelancaster.csv"
}

total_documents = 6  # Assuming 6 documents are being processed

def calculate_weight(freq, max_freq, doc_count, total_documents):
    """Calculate the weight of each token using the provided formula."""
    return (freq / max_freq) * math.log10((total_documents / doc_count) + 1)

def create_descriptor_and_inverser(file_path, descriptor_output, inverser_output):
    # Load the CSV file
    df = pd.read_csv(file_path)
    
    # Calculate max frequency in the document for weighting
    max_frequency = df['Frequency'].max()
    
    # Initialize dictionaries for descriptors and inversers
    descriptor_data = []
    inverser = defaultdict(lambda: {"doc_count": 0, "total_frequency": 0})
    
    for _, row in df.iterrows():
        token = row['Token']
        frequency = row['Frequency']
        doc_id = row['ID']  # Assuming an ID column for document identification
        
        # Update inverser data with document count and frequency per document
        inverser[token]["doc_count"] += 1
        inverser[token]["total_frequency"] += frequency
        
        # Calculate weight for descriptor
        weight = calculate_weight(frequency, max_frequency, inverser[token]["doc_count"], total_documents)
        
        # Append to descriptor data
        descriptor_data.append({
            "Token": token,
            "Frequency": frequency,
            "Number of Documents": total_documents,
            "Weight": weight
        })
    
    # Create descriptor DataFrame and save to CSV
    descriptor_df = pd.DataFrame(descriptor_data)
    descriptor_df.to_csv(descriptor_output, index=False)
    print(f"Descriptor file created at {descriptor_output}")
    
    # Create inverser DataFrame for saving
    inverser_data = {
        "Token": [],
        "Frequency": [],
        "Number of Documents": [],
        "Weight": []
    }
    
    for token, data in inverser.items():
        frequency = data["total_frequency"]
        doc_count = data["doc_count"]
        weight = calculate_weight(frequency, max_frequency, doc_count, total_documents)
        
        inverser_data["Token"].append(token)
        inverser_data["Frequency"].append(frequency)
        inverser_data["Number of Documents"].append(doc_count)
        inverser_data["Weight"].append(weight)
    
    # Save inverser to CSV
    inverser_df = pd.DataFrame(inverser_data)
    inverser_df.to_csv(inverser_output, index=False)
    print(f"Inverser file created at {inverser_output}")

# Run for each preprocessing type
for preprocess_type, file_path in file_paths.items():
    # Set output paths for descriptor and inverser
    descriptor_output = f"data/{preprocess_type}_descriptor.csv"
    inverser_output = f"data/{preprocess_type}_inverser.csv"
    
    # Create descriptor and inverser files
    create_descriptor_and_inverser(file_path, descriptor_output, inverser_output)
```

```python
import os
import pandas as pd
from collections import defaultdict
from nltk.tokenize import RegexpTokenizer
from nltk.stem import PorterStemmer, LancasterStemmer
import math

# Initialize stemmers and tokenizer
porter = PorterStemmer()
lancaster = LancasterStemmer()
regexp_tokenizer = RegexpTokenizer(r'(?:[A-Z]\.)+|\b\w+(?:-\d+(?:\.\d+)?)?\b|\d+(?:\.\d+)?(?:DA)?|\.{3}')

# Define your preprocessing functions
def split_only(text):
    return text.split()

def split_porter(text):
    return [porter.stem(word) for word in text.split()]

def split_lancaster(text):
    return [lancaster.stem(word) for word in text.split()]

def tokenize_only(text):
    return regexp_tokenizer.tokenize(text)

def tokenize_porter(text):
    return [porter.stem(word) for word in regexp_tokenizer.tokenize(text)]

def tokenize_lancaster(text):
    return [lancaster.stem(word) for word in regexp_tokenizer.tokenize(text)]

# Create a dictionary to hold all tokens and their frequencies
all_tokens = defaultdict(lambda: defaultdict(int))
doc_count = defaultdict(int)
max_freq = defaultdict(int)

# Folder containing your text files
folder_path = 'path/to/your/folder'  # Update this to your folder path

# Preprocess each file
for filename in os.listdir(folder_path):
    if filename.endswith('.txt'):
        with open(os.path.join(folder_path, filename), 'r') as file:
            text = file.read()
            tokens = {
                "split_only": split_only(text),
                "split_porter": split_porter(text),
                "split_lancaster": split_lancaster(text),
                "tokenize_only": tokenize_only(text),
                "tokenize_porter": tokenize_porter(text),
                "tokenize_lancaster": tokenize_lancaster(text)
            }

            for method, token_list in tokens.items():
                unique_tokens = set(token_list)
                doc_count[method] += 1  # Count the document for this method
                for token in token_list:
                    all_tokens[method][token] += 1  # Count frequency
                    if all_tokens[method][token] > max_freq[method]:
                        max_freq[method] = all_tokens[method][token]

# Prepare data for descriptors and inversors
descriptors = []
inversors = []

# Calculate weights and prepare descriptors and inversors
for method, token_freq in all_tokens.items():
    for token, freq in token_freq.items():
        num_docs_with_token = sum(1 for v in all_tokens[method].values() if v > 0)
        weight = (freq / max_freq[method]) * math.log10((doc_count[method] / (num_docs_with_token + 1)))  # +1 for log adjustment
        descriptors.append([token, freq, doc_count[method], weight])

# Create DataFrames
descriptors_df = pd.DataFrame(descriptors, columns=['Token', 'Frequency', 'Number of Documents', 'Weight'])
inversors_df = pd.DataFrame(descriptors, columns=['Token', 'Frequency', 'Number of Documents', 'Weight'])  # Similar structure

# Save to CSV
descriptors_df.to_csv('descriptors.csv', index=False)
inversors_df.to_csv('inversors.csv', index=False)

print("Descriptors and Inversors files created successfully!")
```

```python
import os
import pandas as pd
from collections import defaultdict
import nltk
from nltk.tokenize import RegexpTokenizer
from nltk.stem import PorterStemmer, LancasterStemmer
from nltk.corpus import stopwords

nltk.download('stopwords')

# Initialize tokenizers and stemmers
regexp_tokenizer = nltk.RegexpTokenizer(r'(?:[A-Z]\.)+|\d+(?:\.\d+)?(?:DA)?|\w+|\.{3}')
porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
stop_words = set(stopwords.words('english'))
# Paths
input_folder = "Collection"  # Folder with text documents
output_folder = "data2"  # Folder to save descriptor and inverser files

# Function to preprocess text based on the specified method
def preprocess_text(text, method):
    tokens = regexp_tokenizer.tokenize(text)
    tokens = [token for token in tokens if token not in stop_words]
    if "porter" in method:
        tokens = [porter_stemmer.stem(token) for token in tokens]
        

    elif "lancaster" in method:
        tokens = [lancaster_stemmer.stem(token) for token in tokens]

        
    return tokens

# Initialize data structures for all methods
descriptor_data = {method: defaultdict(set) for method in ["split", "splitporter", "splitlancaster", "tokenize", "tokenizeporter", "tokenizelancaster"]}

# Process each file with each method and update descriptor data
for doc_id, file_name in enumerate(os.listdir(input_folder), start=1):
    file_path = os.path.join(input_folder, file_name)
    with open(file_path, 'r', encoding='utf-8') as file:
        text = file.read().lower()
        
        for method in descriptor_data.keys():
            tokens = preprocess_text(text, method)
            unique_tokens = set(tokens)
            
            # Store each token with its associated document ID
            for token in unique_tokens:
                descriptor_data[method][token].add(doc_id)

# Create and save descriptor and inverser files
for method, token_docs in descriptor_data.items():
    # Prepare data for descriptor in the format: doc_id, token (sorted by doc_id and token)
    descriptor_list = [
        {"doc_id": doc_id, "token": token}
        for doc_id in sorted({doc for docs in token_docs.values() for doc in docs})
        for token in sorted(token for token, docs in token_docs.items() if doc_id in docs)
    ]
    
    # Prepare data for inverser in the format: token, doc_id (sorted by doc_id and token)
    inverser_list = [
        {"token": token, "doc_id": doc_id}
        for doc_id in sorted({doc for docs in token_docs.values() for doc in docs})
        for token in sorted(token for token, docs in token_docs.items() if doc_id in docs)
    ]
    
    # Save descriptor and inverser as CSV files
    descriptor_df = pd.DataFrame(descriptor_list)
    inverser_df = pd.DataFrame(inverser_list)
    
    descriptor_df.to_csv(f"{output_folder}/descriptor_{method}.csv", index=False, header=False)
    inverser_df.to_csv(f"{output_folder}/inverser_{method}.csv", index=False, header=False)
    
    print(f"Descriptor and inverser files created for method '{method}'")
```

```python
import pandas as pd
import os

# Define the input and output directories
input_dir = 'data'
output_dir = 'frequency'

# Create the output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# List of descriptor and inverser files to process
descriptor_files = [
    'descriptor_split.csv',
    'descriptor_splitlancaster.csv',
    'descriptor_splitporter.csv',
    'descriptor_tokenize.csv',
    'descriptor_tokenizeporter.csv',
    'descriptor_tokenizelancaster.csv'
]

inverser_files = [
    'inverser_split.csv',
    'inverser_splitlancaster.csv',
    'inverser_splitporter.csv',
    'inverser_tokenize.csv',
    'inverser_tokenizeporter.csv',
    'inverser_tokenizelancaster.csv'
]

# Function to calculate frequencies and max frequencies
def calculate_frequencies(file_path, is_inverser=False):
    # Load the CSV file without headers
    df = pd.read_csv(file_path, header=None)

    # Assign appropriate column names based on the file type
    if is_inverser:
        df.columns = ['Token', 'Document']
    else:
        df.columns = ['Document', 'Token']

    # Clean the token data
    df['Token'] = df['Token'].str.strip().str.lower()

    # Calculate the frequency of each token in each document
    freq_df = df.groupby(['Document', 'Token']).size().reset_index(name='Frequency')

    # Calculate max frequency for each document
    max_freq = freq_df.groupby('Document')['Frequency'].max().reset_index(name='Max Frequency')

    # Merge frequency data with max frequency
    result = pd.merge(freq_df, max_freq, on='Document')

    # Save the results to a new CSV file in the output directory
    result_file = os.path.join(output_dir, f'frequency_{os.path.basename(file_path)}')
    result.to_csv(result_file, index=False)

# Process all descriptor files
for file in descriptor_files:
    file_path = os.path.join(input_dir, file)
    calculate_frequencies(file_path)

# Process all inverser files
for file in inverser_files:
    file_path = os.path.join(input_dir, file)
    calculate_frequencies(file_path, is_inverser=True)

print("Frequency calculation completed for all files.")
```

```python
import os
import pandas as pd
import numpy as np
import math

# Paths to descriptor and inverser files and output folders
descriptor_folder = 'desc'
inverser_folder = 'inv'
output_descriptor_folder = 'data2/desc'
output_inverser_folder = 'data2/inv'

# Make sure output folders exist
os.makedirs(output_descriptor_folder, exist_ok=True)
os.makedirs(output_inverser_folder, exist_ok=True)

# Step 1: Process each descriptor file individually
def process_descriptor_file(file_path, N, n_df):
    df = pd.read_csv(file_path, header=None, names=['doc_number', 'token'])

    # Calculate freq(t,d) for each token in each document
    freq = df.groupby(['doc_number', 'token']).size().reset_index(name='freq')

    # Calculate Max(freq(t,d)) for each document
    freq['max_freq'] = freq.groupby('doc_number')['freq'].transform('max')

    # Merge with `n` DataFrame to get the number of documents containing each token
    result = freq.merge(n_df, on='token', how='left')
    result['N'] = N

    # Calculate log10(N / n + 1)
    result['log_factor'] = np.log10(result['N'] / result['n'] + 1)

    # Calculate poids(t,d)
    result['poids'] = (result['freq'] / result['max_freq']) * result['log_factor']

    return result

# Step 2: Process each inverser file individually
def process_inverser_file(file_path, N):
    df = pd.read_csv(file_path, header=None, names=['token', 'doc_number'])
    
    # Calculate n (number of documents containing each token)
    n = df.groupby('token')['doc_number'].nunique().reset_index(name='n')
    
    # Add N and log_factor
    n['N'] = N
    n['log_factor'] = np.log10(n['N'] / n['n'] + 1)
    
    return n

# Step 3: Main processing loop
def main():
    # First, calculate the total number of documents across all descriptor files
    descriptor_df_list = [pd.read_csv(os.path.join(descriptor_folder, f), header=None, names=['doc_number', 'token'])
                          for f in os.listdir(descriptor_folder) if f.startswith('descriptor')]
    total_descriptor_df = pd.concat(descriptor_df_list, ignore_index=True)
    N = total_descriptor_df['doc_number'].nunique()
    
    # Calculate `n` for each token across all descriptor files (global n for consistency)
    n_df = total_descriptor_df.groupby('token')['doc_number'].nunique().reset_index(name='n')
    
    # Process each descriptor file
    for descriptor_file in os.listdir(descriptor_folder):
        if descriptor_file.startswith('descriptor'):
            file_path = os.path.join(descriptor_folder, descriptor_file)
            
            # Process the descriptor file
            result = process_descriptor_file(file_path, N, n_df)
            
            # Save the processed descriptor file with additional columns
            output_file_path = os.path.join(output_descriptor_folder, f"processed_{descriptor_file}")
            result.to_csv(output_file_path, index=False)
            print(f"Processed descriptor saved to {output_file_path}")
    
    # Process each inverser file
    for inverser_file in os.listdir(inverser_folder):
        if inverser_file.startswith('inverser'):
            file_path = os.path.join(inverser_folder, inverser_file)
            
            # Process the inverser file to calculate `n` values
            result = process_inverser_file(file_path, N)
            
            # Save the processed inverser file with additional columns
            output_file_path = os.path.join(output_inverser_folder, f"processed_{inverser_file}")
            result.to_csv(output_file_path, index=False)
            print(f"Processed inverser saved to {output_file_path}")

# Run the main function
main()
```
