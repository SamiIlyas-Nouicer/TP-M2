```python
import cv2 as cv
import numpy as np

img = cv.imread('pic.jpeg',cv.IMREAD_GRAYSCALE)

laplacien = np.array([[0,-1,0], [-1,4,-1], [0,-1,0]])
gaussien = np.array([[1,2,1], [2,4,2], [1,2,1]]) / 16
kernel = np.array([[0,-1,0], [-1,5,-1], [0,-1,0]])
imgReskernel = cv.filter2D(img.astype(np.int16) ,-1,gaussien)
imgReslaplacien = cv.filter2D(img.astype(np.int16) ,-1,laplacien)
imgRes = cv.filter2D(img.astype(np.int16) ,-1,kernel)

print(imgReslaplacien.dtype)
print(imgReskernel.dtype)


cv.normalize(imgRes, imgRes, 0, 255, cv.NORM_MINMAX)
cv.normalize(imgReslaplacien, imgReslaplacien, 0, 255, cv.NORM_MINMAX)

cv.normalize(imgReskernel, imgReskernel, 0, 255, cv.NORM_MINMAX)
cv.imshow("laplacien",imgReslaplacien.astype(np.uint8))
cv.imshow("kernel",imgRes.astype(np.uint8))

cv.imshow("gaussien",imgReskernel.astype(np.uint8))
cv.imshow("normal",img)
cv.waitKey(0)
cv.destroyAllWindows()
```
