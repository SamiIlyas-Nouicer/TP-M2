```python
import numpy as np
import cv2

def gaussFilter(img, sigma):
    filter = np.zeros((5, 5), dtype=np.float64)
    for i in range(5):
        for j in range(5):
            x = i - 2
            y = j - 2
            filter[i, j] = 1/(2*np.pi*sigma**2)*np.exp(-(x**2+y**2)/(2*sigma**2))
    
    return cv2.filter2D(img, -1, filter)

img = cv2.imread('image.jpg')
imgRes = gaussFilter(img, 1.4)

cv2.imshow("original image", img)
cv2.imshow("Gauss image", imgRes)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
