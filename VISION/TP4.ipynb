```python
import cv2 as cv
import numpy as np

img = cv.imread('image.jpg',cv.IMREAD_GRAYSCALE)
# if img is None:
#     print("Could not open or find the image")
#     exit(0)
# else:    
#     cv.imshow("the pic3",img)
#     cv.waitKey(0)
#     cv.destroyAllWindows()
grad_x = img[:, :img.shape[1]-1] - img[:, 1:]

grad_y = img[:img.shape[0]-1, :] - img[1:, :]
kernel = np.array([[1,2,1], [2,4,2], [1,2,1]]) / 16
grad_x = grad_x[:grad_y.shape[0], :]  
grad_y = grad_y[:, :grad_x.shape[1]] 
carre = cv.filter2D(img ,-1,kernel)

carre = (np.sqrt(grad_x**2 + grad_y**2)
)
cv.normalize(carre, carre, 0, 255, cv.NORM_MINMAX)
carre=  np.uint8(carre *255/np.max(carre))
cv.imshow("the pic after gradient",carre)
cv.waitKey(0)
# cv.destroyAllWindows()
```
