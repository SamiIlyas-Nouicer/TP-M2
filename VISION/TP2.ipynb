```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
# from numpy.lib.shape_base import vsplit
```

```python
img = cv2.imread('pic.jpeg', cv2.IMREAD_GRAYSCALE)
# img = mpimg.imread("pic.jpeg")
image = mpimg.imread('pic2.jpeg')

# img[:] = img[:]/2
imgNorm =np.zeros((img.shape), dtype=np.uint8)
h,w = img.shape
min = 255
max = 0
for i in range(h):
    for j in range(w):
        if img[i,j] > max:
            max = img[i,j]
        if img[i,j] < min:
            min = img[i,j]
            
for i in range(h):
    for j in range(w):
        imgNorm[i,j] = 255*(img[i,j]-min)/(max-min)
print("min", min , "max" ,max)
fig, axs = plt.subplots(1, 3, figsize=(10, 5))

# Display images in each subplot
axs[0].imshow(image)
axs[0].axis('off')  # Hide axis for the first image

axs[1].imshow(imgNorm)
axs[1].axis('off')  # Hide axis for the second image

cv2.imshow('image', img)
cv2.imshow('imageNorm', imgNorm)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
hist1 = np.zeros((256,1), dtype=np.uint16)
for i in range(h):
    for j in range(w):
        hist1[img[i,j]] += 1

hist2 = cv2.calcHist([imgNorm], [0], None, [256], [0,255])

plt.figure()
plt.title('image normalised')
plt.xlabel("grayscale value")
plt.ylabel("nbPixels")
plt.plot(hist2)
plt.plot(hist1)
plt.xlim([0,255])
plt.show()
```
